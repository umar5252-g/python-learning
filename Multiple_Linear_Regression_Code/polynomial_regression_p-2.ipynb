```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-3, 3, 50)
y = x ** 2

plt.scatter(x, y)
plt.xlabel("x")
plt.ylabel("y")
plt.title("currved data")
plt.show()
```

```python
X_linear = np.ones((x.shape[0], 2))
X_linear[:, 1] = x
```

```python
theta_linear = np.linalg.pinv(X_linear.T @ X_linear) @ X_linear.T @ y
y_linear_pred = X_linear @ theta_linear
```

```python
plt.scatter(x, y, label="Actual data")
plt.plot(x, y_linear_pred, color="red", label="Linear model")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()
```

```python
X_poly = np.ones((x.shape[0], 3))
X_poly[:, 1] = x
X_poly[:, 2] = x ** 2
```

```python
theta_poly = np.linalg.pinv(X_poly.T @ X_poly) @ X_poly.T @ y
y_poly_pred = X_poly @ theta_poly
```

```python
plt.scatter(x, y, label="Actual data")
plt.plot(x, y_linear_pred, color="red", label="Linear model")
plt.plot(x, y_poly_pred, color="green", label="Polynomial model")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()
```
