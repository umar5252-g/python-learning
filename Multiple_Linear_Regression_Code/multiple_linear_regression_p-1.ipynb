```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.loadtxt('ex3Data/ex3x.dat', dtype= np.float32)
y = np.loadtxt('ex3Data/ex3y.dat', dtype= np.float32)
```

```python
plt.scatter(x[:, 0], y, marker='x', color='red')
plt.xlabel("House size")
plt.ylabel("House price")
plt.title("House size vs. price")
plt.show()

plt.scatter(x[:, 1], y, marker='x', color='blue')
plt.xlabel("Number of bedrooms")
plt.ylabel("House price")
plt.title("Bedrooms vs. price")
plt.show()
```

```python
X = np.ones((x.shape[0], 3))
X[:, 1] = x[:, 0]
X[:, 2] = x[:, 1]
```

```python
theta = np.linalg.pinv(X.T @ X) @ X.T @ y
theta
```

```python
y_pred = X @ theta
print(y_pred[:5])
```

```python
plt.scatter(y, y_pred)
plt.xlabel("actual house price")
plt.ylabel("predicted house pice")
plt.title("actual vs. predicted prices")
plt.show()
```

```python
error = y_pred - y
rmse = np.sqrt(np.mean(error ** 2))
print ("pridiction error: ", rmse)
```
