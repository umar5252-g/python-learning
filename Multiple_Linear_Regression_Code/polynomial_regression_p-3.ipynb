```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-3, 3, 50)
y = x ** 2

plt.scatter(x, y)
plt.xlabel("x")
plt.ylabel("y")
plt.title("Data for Polynomial Regression")
plt.show()
```

```python
X = np.ones((x.shape[0], 3))
X[:, 1] = x
X[:, 2] = x ** 2
```

```python
theta = np.zeros(3)
learning_rate = 0.01
```

```python
y_pred = X @ theta
error = y_pred - y
loss = np.mean(error ** 2) / 2
print(loss)
```

```python
gradient = X.T @ error / x.shape[0]
print(gradient)
```

```python
theta = theta - learning_rate * gradient
print(theta)
```

```python
theta = np.zeros(3)
loss_history = []

for i in range(50):
    y_pred = X @ theta
    error = y_pred - y
    loss = np.mean(error ** 2) / 2
    gradient = X.T @ error / len(y)
    theta = theta - learning_rate * gradient
    loss_history.append(loss)

print("Final coefficients:", theta)
print("Final loss:", loss_history[-1])
```

```python
plt.plot(loss_history)
plt.xlabel("itteration")
plt.ylabel("loss")
plt.title("loss during Gradient Descent")
plt.show()
```

```python
y_pred = X @ theta

plt.scatter(x, y, label="Actual data")
plt.plot(x, y_pred, color="red", label="Gradient descent fit")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()
```
